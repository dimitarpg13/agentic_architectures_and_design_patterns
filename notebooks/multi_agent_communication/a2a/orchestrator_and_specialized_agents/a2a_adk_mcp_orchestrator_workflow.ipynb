```python
from __future__ import annotations

import sys
from pathlib import Path

notebook_dir = Path.cwd()
if str(notebook_dir) not in sys.path:
    sys.path.insert(0, str(notebook_dir))

from workflow import run_workflow

print(f"Notebook directory: {notebook_dir}")
```

```python
import os

# Use .env defaults, or override here
LLM_PROVIDER = os.getenv("LLM_PROVIDER", "openai")  # "openai" or "anthropic"
OPENAI_API_KEY_INLINE = None
ANTHROPIC_API_KEY_INLINE = None
OPENAI_MODEL = None
ANTHROPIC_MODEL = None

print("Provider:", LLM_PROVIDER)
print("Inline OPENAI key set:", OPENAI_API_KEY_INLINE is not None)
print("Inline ANTHROPIC key set:", ANTHROPIC_API_KEY_INLINE is not None)
```

```python
import asyncio
from pprint import pprint

objective = (
    "Prepare a concise executive brief on using Google ADK and A2A for orchestrated "
    "multi-agent collaboration, and include one KPI-style metric statement."
)

result = await run_workflow(
    objective,
    provider=LLM_PROVIDER,
    openai_api_key=OPENAI_API_KEY_INLINE,
    anthropic_api_key=ANTHROPIC_API_KEY_INLINE,
    openai_model=OPENAI_MODEL,
    anthropic_model=ANTHROPIC_MODEL,
)

print("=== Final Answer ===")
print(result["final_answer"])
```

```python
print("\n=== Plan ===")
pprint(result["plan"])

print("\n=== Step Outputs ===")
pprint(result["outputs"])

print("\n=== A2A Message Trace ===")
pprint(result["a2a_messages"])

print("\nADK orchestrator initialized:", result["adk_orchestrator_initialized"])
```
